## Hacer una lista con nombres propios que inician con mayúscula

```python
nombres = "michel, ramon, yadira, fernando, joselin, andre, rodolfo, nairo, mitsue, laura, luisa, turbo, memo"
```

```python
nombres_l=nombres.replace(", ",",").title().split(",")
```

```python
for n in nombres_l:
    if n.endswith("o"):
        print(n)
```
